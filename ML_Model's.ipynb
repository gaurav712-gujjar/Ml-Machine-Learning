```python
# (1). LinearRegression Model --- > when our target data is numerical .
```

```python
import pandas as pd
```

```python
df  = pd.read_csv("/content/insurance - insurance.csv")
df.head(3)
```

```python
x = df.drop(columns=['charges'])
y = df['charges']
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
```

```python
x['sex']=le.fit_transform(x['sex'])
x['smoker']=le.fit_transform(x['smoker'])
x['region']=le.fit_transform(x['region'])
```

```python
x.head(3)
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.2, random_state=42)
```

```python
# Steps-3 ---> select ml model (LinearRegression)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
```

```python
lr.fit(x_train, y_train)
```

```python
y_pred = lr.predict(x_test)
```

```python
from sklearn.metrics import r2_score

# r2_score ---> if data is numerical
# Accuracy Score ---> if data is categorial
```

```python
r2_score(y_test, y_pred)
```

**   STDANDARD_SCALAR**

StandardScaler is used to standardize numerical data so that all features are on the same scale.

```python
# for converting our data into normal distribution we will use standardscaler
```

```python
from sklearn.preprocessing import StandardScaler
# StandardScaler rescales numbers so all features are on the same scale,
# without changing their distribution shape(data).
# Ensures all features are on the same scale
```

```python
sc = StandardScaler()
```

```python
x_train_sc = sc.fit_transform(x_train)    ## fit--> perform operatoin mean , standard deviation
                                          ## transform --> apply the formula to all data
```

```python
x_train_new = pd.DataFrame(x_train_sc, columns = x_train.columns)  #convert into dataframe for use numpy round function
```

```python
np.round(x_train_new.describe(), 2)
```
